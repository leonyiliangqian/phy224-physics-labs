```python
import numpy as np
import pandas as pd
from uncertainties import unumpy as unp
from uncertainties import ufloat
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
def Iunc(I):
    if I <= 30:
        return .2/100 * I + .005
    
    return .2/100 * I + .05

def model(V, Is, VT):
    return Is * (np.exp(V/VT) - 1)
```

```python
# Read data
data = pd.read_csv('Diode_positive_resistor.csv')
voltage = data['Voltage (V)']
current = data['Current (mA)']

# Convert data to uncertainties
current = unp.uarray(current, [Iunc(I) for I in current])
```

```python
# Plot the data
plt.plot(voltage, unp.nominal_values(current), 'o', label='Data')
plt.xlabel('Voltage (V)')
plt.ylabel('Current (mA)')
plt.title('Diode With Resistor I-V Curve')
plt.legend()
plt.grid()
plt.show()
```
